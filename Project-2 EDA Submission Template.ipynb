#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
plt.figure(figsize=(8,5))

# Fill NaN values in 'character' with 'Unknown' for this crosstab
# This ensures that 'Unknown' values are properly categorized.
character_filled = df['character'].fillna('Unknown')

# Create the crosstab based on whether the character is 'Unknown' or not
ct = pd.crosstab(df['role'], character_filled == 'Unknown')

# Rename the columns to be more descriptive
# The columns of ct will be `False` (for known characters) and `True` (for unknown characters).
# We are guaranteed to have both columns because df.info() shows NaNs in 'character',
# which after fillna will become 'Unknown', and there are also many known characters.
ct.columns = ['Known Character', 'Unknown Character']

sns.heatmap(ct,annot=True,fmt='d',cmap='Blues')
plt.title('Heatmap: Role vs Character Availability')
plt.show()
````
